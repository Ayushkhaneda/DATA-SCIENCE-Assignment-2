```python
import pandas as pd
import numpy as np
import ast
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

df = pd.read_csv("imdb_data.csv")

df['genres'] = df['genres'].fillna("[]")
df['cast'] = df['cast'].fillna("[]")
df['crew'] = df['crew'].fillna("[]")

df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')

df['profit'] = df['revenue'] - df['budget']
df['roi'] = df['profit'] / df['budget'].replace(0, np.nan)

def extract_names(data, job):
    names = []
        try:
                items = ast.literal_eval(data)
                        for i in items:
                                    if 'job' in i and i['job'] == job:
                                                    names.append(i['name'])
                                                        except:
                                                                pass
                                                                    return names

                                                                    def extract_cast(data):
                                                                        names = []
                                                                            try:
                                                                                    items = ast.literal_eval(data)
                                                                                            for i in items:
                                                                                                        names.append(i['name'])
                                                                                                            except:
                                                                                                                    pass
                                                                                                                        return names

                                                                                                                        def extract_genres(data):
                                                                                                                            names = []
                                                                                                                                try:
                                                                                                                                        items = ast.literal_eval(data)
                                                                                                                                                for i in items:
                                                                                                                                                            names.append(i['name'])
                                                                                                                                                                except:
                                                                                                                                                                        pass
                                                                                                                                                                            return names

                                                                                                                                                                            max_profit = df['profit'].max()
                                                                                                                                                                            max_movie = df[df['profit'] == max_profit].iloc[0]

                                                                                                                                                                            title = max_movie['title']
                                                                                                                                                                            profit = max_movie['profit']
                                                                                                                                                                            directors = extract_names(max_movie['crew'], 'Director')
                                                                                                                                                                            producers = extract_names(max_movie['crew'], 'Producer')
                                                                                                                                                                            cast = extract_cast(max_movie['cast'])

                                                                                                                                                                            print("Highest Profit Movie:")
                                                                                                                                                                            print("Title:", title)
                                                                                                                                                                            print("Profit:", profit)
                                                                                                                                                                            print("Directors:", directors)
                                                                                                                                                                            print("Producers:", producers)
                                                                                                                                                                            print("Cast:", cast[:5])

                                                                                                                                                                            language_group = df.groupby('original_language')
                                                                                                                                                                            roi_by_language = language_group['roi'].mean()
                                                                                                                                                                            roi_sorted = roi_by_language.sort_values(ascending=False)

                                                                                                                                                                            print("\nTop Languages by ROI:")
                                                                                                                                                                            print(roi_sorted.head(5))

                                                                                                                                                                            all_genres = set()

                                                                                                                                                                            for g in df['genres']:
                                                                                                                                                                                genres_list = extract_genres(g)
                                                                                                                                                                                    for genre in genres_list:
                                                                                                                                                                                            all_genres.add(genre)

                                                                                                                                                                                            print("\nUnique Genres:")
                                                                                                                                                                                            print(all_genres)

                                                                                                                                                                                            producer_list = []
                                                                                                                                                                                            roi_list = []

                                                                                                                                                                                            for index, row in df.iterrows():
                                                                                                                                                                                                crew_data = row['crew']
                                                                                                                                                                                                    roi_value = row['roi']
                                                                                                                                                                                                        producers = extract_names(crew_data, 'Producer')
                                                                                                                                                                                                            for p in producers:
                                                                                                                                                                                                                    producer_list.append(p)
                                                                                                                                                                                                                            roi_list.append(roi_value)

                                                                                                                                                                                                                            producer_df = pd.DataFrame({
                                                                                                                                                                                                                                'producer': producer_list,
                                                                                                                                                                                                                                    'roi': roi_list
                                                                                                                                                                                                                                    })

                                                                                                                                                                                                                                    producer_group = producer_df.groupby('producer')
                                                                                                                                                                                                                                    producer_roi = producer_group['roi'].mean()
                                                                                                                                                                                                                                    top_producers = producer_roi.sort_values(ascending=False).head(3)

                                                                                                                                                                                                                                    print("\nTop Producers by ROI:")
                                                                                                                                                                                                                                    print(top_producers)

                                                                                                                                                                                                                                    actor_count = {}

                                                                                                                                                                                                                                    for c in df['cast']:
                                                                                                                                                                                                                                        cast_list = extract_cast(c)
                                                                                                                                                                                                                                            for actor in cast_list:
                                                                                                                                                                                                                                                    if actor in actor_count:
                                                                                                                                                                                                                                                                actor_count[actor] += 1
                                                                                                                                                                                                                                                                        else:
                                                                                                                                                                                                                                                                                    actor_count[actor] = 1

                                                                                                                                                                                                                                                                                    top_actor = max(actor_count, key=actor_count.get)
                                                                                                                                                                                                                                                                                    top_actor_count = actor_count[top_actor]

                                                                                                                                                                                                                                                                                    print("\nMost Frequent Actor:")
                                                                                                                                                                                                                                                                                    print(top_actor, top_actor_count)

                                                                                                                                                                                                                                                                                    actor_movies = []

                                                                                                                                                                                                                                                                                    for index, row in df.iterrows():
                                                                                                                                                                                                                                                                                        cast_list = extract_cast(row['cast'])
                                                                                                                                                                                                                                                                                            if top_actor in cast_list:
                                                                                                                                                                                                                                                                                                    actor_movies.append({
                                                                                                                                                                                                                                                                                                                'title': row['title'],
                                                                                                                                                                                                                                                                                                                            'profit': row['profit'],
                                                                                                                                                                                                                                                                                                                                        'genres': extract_genres(row['genres'])
                                                                                                                                                                                                                                                                                                                                                })

                                                                                                                                                                                                                                                                                                                                                actor_movies_df = pd.DataFrame(actor_movies)

                                                                                                                                                                                                                                                                                                                                                print("\nSample Movies of Top Actor:")
                                                                                                                                                                                                                                                                                                                                                print(actor_movies_df.head())

                                                                                                                                                                                                                                                                                                                                                director_actor_map = {}

                                                                                                                                                                                                                                                                                                                                                for index, row in df.iterrows():
                                                                                                                                                                                                                                                                                                                                                    directors = extract_names(row['crew'], 'Director')
                                                                                                                                                                                                                                                                                                                                                        cast_list = extract_cast(row['cast'])
                                                                                                                                                                                                                                                                                                                                                            for d in directors:
                                                                                                                                                                                                                                                                                                                                                                    if d not in director_actor_map:
                                                                                                                                                                                                                                                                                                                                                                                director_actor_map[d] = []
                                                                                                                                                                                                                                                                                                                                                                                        director_actor_map[d].extend(cast_list)

                                                                                                                                                                                                                                                                                                                                                                                        all_directors = []

                                                                                                                                                                                                                                                                                                                                                                                        for c in df['crew']:
                                                                                                                                                                                                                                                                                                                                                                                            directors = extract_names(c, 'Director')
                                                                                                                                                                                                                                                                                                                                                                                                for d in directors:
                                                                                                                                                                                                                                                                                                                                                                                                        all_directors.append(d)

                                                                                                                                                                                                                                                                                                                                                                                                        director_series = pd.Series(all_directors)
                                                                                                                                                                                                                                                                                                                                                                                                        top_directors = director_series.value_counts().head(3)

                                                                                                                                                                                                                                                                                                                                                                                                        print("\nTop Directors:")
                                                                                                                                                                                                                                                                                                                                                                                                        print(top_directors)

                                                                                                                                                                                                                                                                                                                                                                                                        print("\nDirector Preferences:")

                                                                                                                                                                                                                                                                                                                                                                                                        for d in top_directors.index:
                                                                                                                                                                                                                                                                                                                                                                                                            actors = director_actor_map[d]
                                                                                                                                                                                                                                                                                                                                                                                                                actor_series = pd.Series(actors)
                                                                                                                                                                                                                                                                                                                                                                                                                    top_actors = actor_series.value_counts().head(3)
                                                                                                                                                                                                                                                                                                                                                                                                                        print("\nDirector:", d)
                                                                                                                                                                                                                                                                                                                                                                                                                            print(top_actors)

                                                                                                                                                                                                                                                                                                                                                                                                                            language_roi = df.groupby('original_language')['roi'].mean()
                                                                                                                                                                                                                                                                                                                                                                                                                            top_languages = language_roi.sort_values(ascending=False).head(10)

                                                                                                                                                                                                                                                                                                                                                                                                                            plt.figure()
                                                                                                                                                                                                                                                                                                                                                                                                                            top_languages.plot(kind='bar')
                                                                                                                                                                                                                                                                                                                                                                                                                            plt.title("Top 10 Languages by ROI")
                                                                                                                                                                                                                                                                                                                                                                                                                            plt.xlabel("Language")
                                                                                                                                                                                                                                                                                                                                                                                                                            plt.ylabel("ROI")
                                                                                                                                                                                                                                                                                                                                                                                                                            plt.show()
```
